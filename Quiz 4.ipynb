## Question 1  
**How often is `f(x)` called?**

```python
f_lam = lambda x: x + 1
print(f_lam(5))
```

```python
def f_non_lambda(x):
    return x + 1

print(f_non_lambda(5))
```

```python
add = lambda a, b: a + b
print(add(3, 4))
```

```python
print((lambda x: x * 2)(5))
```

```python
nums = [1, 2, 3, 4]
result = list(map(lambda x: x * 2, nums))
print(result)
```

```python
nums = [1, 2, 3, 4, 5, 6]
result = list(filter(lambda x: x % 2 == 0, nums))
print(result)
```

```python
f = lambda x: "even" if x % 2 == 0 else "odd"
print(f(3))
print(f(4))
```

```python
def f(x):
    print(f"f({x}) called")
    return x + 2

result = list(map(f, filter(lambda x: f(x) % 2 == 0, range(4))))
print("Final result:", result)
```

### Explanation

`range(4)` produces `0, 1, 2, 3`.

The `filter(...)` part calls `f(x)` once for each of these 4 values to test whether `f(x) % 2 == 0`.

Only the values that pass the filter are then sent to `map(f, ...)`, which calls `f(x)` again on those surviving values.

Since `f(x) = x + 2`, the values are:

- `f(0) = 2` → even → kept
- `f(1) = 3` → odd → removed
- `f(2) = 4` → even → kept
- `f(3) = 5` → odd → removed

So `f` is called:

- 4 times inside `filter`
- 2 more times inside `map`

**Total: 6**

## Question 2  
**Sorting digits in descending order and converting back to integers**

```python
to_sort = [1, 10, 23, 33, 42, 100, 48279]

sorted_list = list(map(lambda d: int("".join(sorted(str(d), reverse=True))), to_sort))
print(sorted_list)
```

### Explanation

This program **does work exactly as intended**.

For each number:

1. `str(d)` converts the number to a string
2. `sorted(..., reverse=True)` sorts the characters in descending order
3. `"".join(...)` combines them back into one string
4. `int(...)` converts that string back into an integer

Example:

- `48279` → `"48279"`
- sorted descending → `['9', '8', '7', '4', '2']`
- joined → `"98742"`
- converted → `98742`

So the correct statement is:

**The assumption that the program would not do what it is supposed to do is incorrect. In fact, the program fulfills the requirements exactly.**

## Question 4  
**Recursive Fibonacci style function: which statements are true?**

```python
def fibonacci(n):
    n = int(n)
    if n == 0 or n == 1:
        return n
    return fibonacci(n - 2) + fibonacci(n - 1)

print(fibonacci(10))
```

```python

fibonacci(-1)
```

### Explanation

True statement:

- **`fibonacci(n)` is a recursive function.**

Why the others are false:

- `fibonacci(1)` is `1`, but `fibonacci(-1)` does not reach a base case and would recurse forever until Python raises a `RecursionError`
- Having two base cases is normal and does **not** make the function invalid
- Python can still execute the function for negative integers, but it will not terminate correctly
- `fibonacci(1) = 1`, while `fibonacci(0) = 0`, so they are not the same

## Question 6  
**Given `recursive_sum(number)`, which statements are true?**

```python
def recursive_sum(number):
    number = int(number)

    if number < 0:
        raise ValueError(
            f'recursive_sum requires a non-negative argument but has been called with number={number}'
        )
    elif number == 1:
        return 1
    else:
        return number + recursive_sum(number - 1)
```

```python
print("recursive_sum(1.1234) =", recursive_sum(1.1234))
print("recursive_sum(4) =", recursive_sum(4))

try:
    recursive_sum(0)
except Exception as e:
    print("recursive_sum(0) raised:", type(e).__name__, str(e))

print("recursive_sum(2.1234) =", recursive_sum(2.1234))
```

### Explanation

The function first converts the input with `int(number)`.

That means:

- `recursive_sum(1.1234)` becomes `recursive_sum(1)` → returns `1`
- `recursive_sum(2.1234)` becomes `recursive_sum(2)` → returns `2 + recursive_sum(1) = 3`, **not** `2`
- `recursive_sum(4)` returns `4 + 3 + 2 + 1 = 10`

Important edge case:

- `recursive_sum(0)` does **not** hit the base case `number == 1`
- it becomes `0 + recursive_sum(-1)`
- then `recursive_sum(-1)` raises a **ValueError**
- so it does **not** raise a `RecursionError`

### True statements

- `recursive_sum` is a recursive function
- `recursive_sum(1.1234)` returns `1`
- the function can be executed with all positive integers greater than `0` without error
- `recursive_sum(4)` returns `10`
- the function can run without problems even though it contains multiple `return` statements

### False statements

- it does **not** return the sum of only even numbers
- it does **not** return the sum of only odd numbers
- `recursive_sum(0)` does **not** cause a `RecursionError`; it causes a `ValueError`
- not all floats between `1` and `101` work without error in practice, because sufficiently large values can still exceed Python's recursion limit after conversion
- `recursive_sum(2.1234)` does **not** return `2`; it returns `3`
