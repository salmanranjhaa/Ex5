# Introduction to Object Oriented Programming with a Ferrari Formula 1 Example

This notebook explains the core ideas of Object Oriented Programming using a **Ferrari Formula 1 car** instead of the Mario Kart example.

## 1. Intuition: why OOP?

Imagine Ferrari wants to model many Formula 1 cars in software.

Each car has:

- a chassis
- an engine mode
- tyres
- a top speed
- a current speed

Each car can also **do things**:

- accelerate
- brake
- pit
- change tyres
- report its status

Instead of rewriting the same logic for every car, we define a **class** as a template.

Then we create actual **objects** from that template.

## 2. A first class: `FerrariF1Car`

A class defines:

- **attributes**: the data an object stores
- **methods**: the actions an object can perform

````python
class FerrariF1Car:
    def __init__(self, driver, chassis, tyre_compound="soft"):
        self.driver = driver
        self.chassis = chassis
        self.tyre_compound = tyre_compound
        self.speed = 0
        self.top_speed = 350
        self.in_pit = False

    def accelerate(self, amount):
        self.speed = min(self.speed + amount, self.top_speed)

    def brake(self, amount):
        self.speed = max(self.speed - amount, 0)

    def pit(self):
        self.in_pit = True
        self.speed = 80

    def exit_pit(self):
        self.in_pit = False

    def status(self):
        return {
            "driver": self.driver,
            "chassis": self.chassis,
            "tyre_compound": self.tyre_compound,
            "speed": self.speed,
            "top_speed": self.top_speed,
            "in_pit": self.in_pit
        }
````

### Explanation

In this class:

**Attributes**
- `driver`
- `chassis`
- `tyre_compound`
- `speed`
- `top_speed`
- `in_pit`

**Methods**
- `accelerate`
- `brake`
- `pit`
- `exit_pit`
- `status`

So the class says what a Ferrari F1 car **has** and what it **can do**.

## 3. Creating objects

A class is just the blueprint.

An **object** is a concrete instance of that class.

````python
car1 = FerrariF1Car("Charles Leclerc", "SF 24", "soft")
car2 = FerrariF1Car("Lewis Hamilton", "SF 24", "medium")

print(car1.status())
print(car2.status())
````

### Explanation

`car1` and `car2` are two different objects.

They are both Ferrari F1 cars, but they hold different values.

This is the core class/object idea:

- **class** = template
- **object** = one real instance created from that template

## 4. Accessing attributes and calling methods

To access an attribute:

```python
object_name.attribute_name
```

To call a method:

```python
object_name.method_name(arguments)
```

````python
print("Driver:", car1.driver)
print("Tyres:", car1.tyre_compound)

car1.accelerate(120)
print("Speed after accelerate:", car1.speed)

car1.brake(40)
print("Speed after brake:", car1.speed)
````

### Explanation

This is standard dot syntax:

- `car1.driver` reads the attribute
- `car1.accelerate(120)` calls a method

Notice that we do **not** write `self` when calling the method from the object.

Inside the class definition, Python handles that automatically.

## 5. The `__init__` method

`__init__` is a special method that runs when an object is created.

It is used to initialise the object's attributes.

````python
car3 = FerrariF1Car("The Bear", "Future Ferrari", "hard")
print(car3.status())
````

### Explanation

When Python sees:

```python
car3 = FerrariF1Car("The Bear", "Future Ferrari", "hard")
```

it automatically calls:

```python
__init__(self, "The Bear", "Future Ferrari", "hard")
```

That is why `__init__` is often called the constructor.

## 6. Making objects easier to print with `__str__`

Without `__str__`, printing an object is not very informative.

Let's improve that.

````python
class FerrariF1Car:
    def __init__(self, driver, chassis, tyre_compound="soft"):
        self.driver = driver
        self.chassis = chassis
        self.tyre_compound = tyre_compound
        self.speed = 0
        self.top_speed = 350
        self.in_pit = False

    def accelerate(self, amount):
        self.speed = min(self.speed + amount, self.top_speed)

    def brake(self, amount):
        self.speed = max(self.speed - amount, 0)

    def pit(self):
        self.in_pit = True
        self.speed = 80

    def exit_pit(self):
        self.in_pit = False

    def __str__(self):
        return f"Ferrari F1 car driven by {self.driver}, chassis {self.chassis}, tyres {self.tyre_compound}, speed {self.speed} km/h"
````

````python
car1 = FerrariF1Car("Charles Leclerc", "SF 24", "soft")
print(car1)
````

### Explanation

`__str__` returns a human readable string.

It is used automatically when you do:

```python
print(car1)
```

This makes objects much easier to present.

## 7. `__repr__` for debugging style output

`__repr__` is supposed to be a more exact representation of the object.

````python
class FerrariF1Car:
    def __init__(self, driver, chassis, tyre_compound="soft"):
        self.driver = driver
        self.chassis = chassis
        self.tyre_compound = tyre_compound
        self.speed = 0
        self.top_speed = 350
        self.in_pit = False

    def accelerate(self, amount):
        self.speed = min(self.speed + amount, self.top_speed)

    def brake(self, amount):
        self.speed = max(self.speed - amount, 0)

    def pit(self):
        self.in_pit = True
        self.speed = 80

    def exit_pit(self):
        self.in_pit = False

    def __str__(self):
        return f"{self.driver} in {self.chassis} on {self.tyre_compound} tyres at {self.speed} km/h"

    def __repr__(self):
        return f"FerrariF1Car(driver={self.driver!r}, chassis={self.chassis!r}, tyre_compound={self.tyre_compound!r}, speed={self.speed!r})"
````

````python
car1 = FerrariF1Car("Charles Leclerc", "SF 24", "soft")
print(str(car1))
print(repr(car1))
````

### Explanation

- `__str__` is for readable output
- `__repr__` is for a more precise representation, often useful for debugging

## 8. Inheritance: making a specialised Ferrari subclass

Suppose we want a more specific car type for qualifying.

A qualifying car is still a Ferrari F1 car, but with slightly different behaviour.

This is where **inheritance** comes in.

````python
class FerrariQualifyingCar(FerrariF1Car):
    def __init__(self, driver, chassis, tyre_compound="soft"):
        super().__init__(driver, chassis, tyre_compound)
        self.top_speed = 360
        self.ers_mode = "qualifying"

    def deploy_ers(self):
        self.speed = min(self.speed + 15, self.top_speed)
````

````python
quali_car = FerrariQualifyingCar("Charles Leclerc", "SF 24 Q", "soft")
print(quali_car)
print("Top speed:", quali_car.top_speed)
print("ERS mode:", quali_car.ers_mode)
````

````python
practice_car = FerrariF1Car("The Bear", "SF 24 P", "medium")
print("Top speed:", practice_car.top_speed)
print("ERS mode:", practice_car.ers_mode)
````

````python
print("normal has ers_mode:", hasattr(practice_car, "ers_mode"))
print("quali has ers_mode:", hasattr(quali_car, "ers_mode"))
````

### Explanation

`FerrariQualifyingCar` is a **subclass** of `FerrariF1Car`.

That means it inherits the attributes and methods of the parent class, and then adds or changes some features.

`super().__init__(...)` reuses the parent class setup instead of rewriting it.

## 9. Inherited methods still work

A subclass automatically gets methods from the parent unless we override them.

````python
quali_car.accelerate(200)
print(quali_car)

quali_car.deploy_ers()
print("After ERS deployment:", quali_car)
````

### Explanation

Even though `deploy_ers()` only exists in the subclass, `accelerate()` still works because it was inherited from the parent class.

## 10. Polymorphism: overriding methods

Now let us change how braking works for the qualifying car.

This is an example of **polymorphism** through method overriding.

````python
class FerrariQualifyingCar(FerrariF1Car):
    def __init__(self, driver, chassis, tyre_compound="soft"):
        super().__init__(driver, chassis, tyre_compound)
        self.top_speed = 360
        self.ers_mode = "qualifying"

    def deploy_ers(self):
        self.speed = min(self.speed + 15, self.top_speed)

    def brake(self, amount):
        reduced_amount = amount * 0.8
        self.speed = max(self.speed - reduced_amount, 0)
````

````python
normal_car = FerrariF1Car("Carlos Sainz", "SF 24", "medium")
quali_car = FerrariQualifyingCar("Charles Leclerc", "SF 24 Q", "soft")

normal_car.accelerate(200)
quali_car.accelerate(200)

normal_car.brake(50)
quali_car.brake(50)

print("Normal car speed after braking:", normal_car.speed)
print("Qualifying car speed after braking:", quali_car.speed)
````

### Explanation

Both classes have a method named `brake`, but the subclass gives it different behaviour.

That is the idea of polymorphism here:

same method name, different implementation

## 11. Accessing attributes directly

By default, Python lets you read and overwrite attributes directly.

````python
car = FerrariF1Car("Charles Leclerc", "SF 24", "soft")
print("Initial speed:", car.speed)

car.speed = 310
print("Updated speed:", car.speed)
````

### Explanation

Direct access is simple, but it can also be risky.

For example, what if someone sets:

```python
car.speed = -999
```

That may be logically wrong for the problem you are modelling.

## 12. `_name` and `__name` in simple words

Underscores in Python are mostly **signals** about how an attribute should be used.

- `car.speed` means a normal public attribute
- `car._engine_temperature` means "this is internal, please do not touch it unless you have a reason"
- `car.__radio_code` means Python will rename it internally to make accidental access harder


````python
class FerrariUnderscoreDemo:
    def __init__(self):
        self.speed = 0
        self._engine_temperature = 95
        self.__radio_code = "BOX, BOX"

car = FerrariUnderscoreDemo()

print("Public attribute:", car.speed)
print("Single underscore still works:", car._engine_temperature)

try:
    print(car.__radio_code)
except AttributeError:
    print("Double underscore cannot be accessed with car.__radio_code")

print("Python stored it as:", car._FerrariUnderscoreDemo__radio_code)
````

### Explanation

A **single underscore** does not block access. It is just a convention that says: "this belongs to the inside of the class".

A **double underscore** is stronger. Python changes the name internally to something like `_FerrariUnderscoreDemo__radio_code`. This is called **name mangling**.

So `__name` is not true security, but it does help prevent accidental misuse.


## 13. Getters and setters

If direct access is risky, we can put methods in the middle.

- a **getter** reads a value
- a **setter** changes a value after checking it

This is helpful when a Ferrari attribute should stay within a sensible range.


````python
class FerrariControlled:
    def __init__(self, engine_temperature=90):
        self._engine_temperature = 90
        self.set_engine_temperature(engine_temperature)

    def get_engine_temperature(self):
        return self._engine_temperature

    def set_engine_temperature(self, value):
        if value < 0:
            raise ValueError("Engine temperature cannot be negative")
        if value > 200:
            raise ValueError("Engine temperature is unrealistically high")
        self._engine_temperature = value
````

````python
car = FerrariControlled()
print("Initial temperature:", car.get_engine_temperature())

car.set_engine_temperature(110)
print("Updated temperature:", car.get_engine_temperature())

try:
    car.set_engine_temperature(500)
except ValueError as e:
    print("Error:", e)
````

### Explanation

Now every update goes through `set_engine_temperature()`.

That means the class can reject bad values before saving them.

This is safer than direct access, but the syntax is a bit more verbose because we need `get_...()` and `set_...()` methods.


## 14. A cleaner Python style: `@property`

Python has a more natural way to write getters and setters.

With `@property`, we still write `car.engine_temperature`, but Python quietly runs methods in the background.


````python
class FerrariPropertyControlled:
    def __init__(self, engine_temperature=90):
        self.engine_temperature = engine_temperature

    @property
    def engine_temperature(self):
        return self._engine_temperature

    @engine_temperature.setter
    def engine_temperature(self, value):
        if value < 0:
            raise ValueError("Engine temperature cannot be negative")
        if value > 200:
            raise ValueError("Engine temperature is unrealistically high")
        self._engine_temperature = value


car = FerrariPropertyControlled()
print("Initial temperature:", car.engine_temperature)

car.engine_temperature = 110
print("Updated temperature:", car.engine_temperature)

try:
    car.engine_temperature = 500
except ValueError as e:
    print("Error:", e)
````

### Explanation

From the outside, `car.engine_temperature` looks like a normal attribute.

Inside the class, Python is really calling the getter or setter method for us.

This is why `@property` is often the most Pythonic option: clean syntax for the user, validation logic for the class.
